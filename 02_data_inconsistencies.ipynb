# Dealing with Inconsistencies in the DataFrame

```python
# import librares
import pandas as pd
import numpy as np
```

```python
data = pd.DataFrame({
    "Date" : ["2021-12-01", "2021-12-02", "2021-12-03", "2021-12-04", "2021-12-05"],
    "Name" : ["Alice", "Alcie", "Charlie", "Eve", "Eve"],
    "Country" : ["USA", "U.S.A", "United States", "United States of America", "U-S-A"],
    "Sales 2020" : [1000, 1500, 2000, np.nan, 3000],
    "Sales 2021" : [1200, 1800, 2200, 2700, np.nan]
    })
```

```python
data.head()
```

```python
# homonize the name of the country 
customize_country = {"USA" : "United States", "U.S.A" : "United States", "United States of America" : "United States", "U-S-A" : "United States"}
data['Country'] = data['Country'].replace(customize_country)
data.head()
```

```python
# correct the name
data['Name'] = data['Name'].replace({"Alcie" : "Alice"})
data.head()
```

```python
# remove duplicates
data = data.drop_duplicates(subset="Name")
data.head()
```
