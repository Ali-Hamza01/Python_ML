# Data Merging

```python
# import libraries

import pandas as pd

# dataset 01

data1 = {
    "Id" : [1, 2, 3],
    "Name" : ['Ali', 'Fatima', 'Ahmad'],
    "Age" : ['35', '28', '27']
}

df1 = pd.DataFrame(data1)

# dataset 02
data2 = {
    "Id" : [1, 2, 3],
    "City" : ['Karachi', 'Faisalabad', 'Lahore'],
    "Occupation" : ['Doctor', 'Teacher', 'Engineer'] 
}

df2 = pd.DataFrame(data2)

# merging the data
merged_data = pd.merge(df1, df2, on='Id', how='inner')
merged_data
```

# Concatenation

```python
data3 ={
    "Name" : ['Usama', 'Mahnoor', "Iqra"],
    "Age" : [ 19, 20, 21]
}

data4 ={
    "Name" : ['Waheed', 'Asif', 'Aslam'],
    "Age" : [24, 20, 19]
}

data5 = {
    "City" : ['Bawalpur', 'Islamabad', 'Pindi'],
    "Height" : [5.11, 6.2, 5.7]
}

df3 = pd.DataFrame(data3)
df4 = pd.DataFrame(data4)
df5 = pd.DataFrame(data5)

# concatenate the data with axis=0
concatenate_data = pd.concat([df3, df4])
print(concatenate_data)

print("--------------------------------")
#concatenate the data with axis=1
concatenate_data2 = pd.concat([df4, df5], axis=1)
print(concatenate_data2)
```
