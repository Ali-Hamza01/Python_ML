# 01-Standard Scaling

```python
# import libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler

# make a dumy dataset
data = {
    "Age" : [10, 24, 25, 26, 50],
    "Height" : [5.11, 5.12, 6.01, 6.02, 6.03],
    "Weight" : [56, 60, 69, 72, 59]
}

df = pd.DataFrame(data)
df
```

```python
# impport StandardScalar
scalar = StandardScaler()

Scalad_df = scalar.fit_transform(df)
Scalad_df
```

# MinMax Scalar


```python
# Transform the data using MinMaxScaler
Scalar = MinMaxScaler()

Scalad_df = Scalar.fit_transform(df)
Scalad_df
```

# MaxABS Scalar

```python
# trasnform the data using MaxAbsScaler
Scalar = MaxAbsScaler()

Scalad_df = Scalar.fit_transform(df)
Scalad_df
```
