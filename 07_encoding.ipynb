# Feature Encoding

Feature Encoding is the proccess of transforming `categorical features` into `numeric features`.Ths is necessary bcz ML algorithms can only handle numeric features such as:

- Label Encoding
- Ordinal Encoding
- One-hot Encoding
- Binary Encoding


```python
# import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = sns.load_dataset('tips')
data.head()
```

```python
# value_counts of time
data['time'].value_counts()
```

## 01-Label Encoding


```python
# now encode the time column using label encoding
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
le = LabelEncoder()

data['encoded_time'] = le.fit_transform(data['time'])
data.tail(20)
```

## 02-Ordinal Encoding

```python
# check the day column value counts
data['day'].value_counts()
```

```python
# now encode the day column using ordinal encoding
oe = OrdinalEncoder(categories=[['Thur', 'Fri', 'Sat', 'Sun']])
data['encoded_day']= oe.fit_transform(data[['day']])
data.tail()
```

## 03- One-Hot Encoding

```python
ohe = OneHotEncoder()
ohe.fit_transform(data[['sex']]).toarray()
```
